# Pandas Tutorial

Pandas is the most widely used data analysis and manipulation library in Python. It builds on top of NumPy to provide labeled, tabular data structures, the `Series` for a single labeled column and the `DataFrame` for a full table, similar to a spreadsheet or a SQL table.

Pandas is used to load data from files, clean and filter it, transform it by adding or reshaping columns, and summarize it with grouping and aggregation. Under the hood, DataFrame columns are usually stored as NumPy arrays, and you can freely convert between the two with `.to_numpy()` or by building a DataFrame from an array.

## Importing pandas and checking the version

Install pandas with `pip install pandas`. It is conventionally imported as `pd`, and NumPy is often imported alongside it since the two are used together so often.

```python
import pandas as pd
import numpy as np

print("pandas version:", pd.__version__)
```

## Series, DataFrame, and concatenating data

A `Series` is a one dimensional labeled array, similar to a single spreadsheet column, with an index of labels alongside the values. A `DataFrame` is a two dimensional table, essentially a collection of Series that share the same index, and is the structure you will use most often.

`pd.concat()` combines multiple Series or DataFrames into one. By default it stacks them as new rows; pass `axis=1` to line them up side by side as new columns instead.

```python
s = pd.Series([10, 20, 30, 40], index=['1', 'b', 'c', 'd'])
print(s)
print("Value at 'b':", s['b'])

# a,b,c,d
# 10,11,12,13 - p1
# 10,23,45,2 - p2
```

```python
data = {
    'Name': ['Alice', 'Bob', 'Charlie', 'Diana'],
    'Age': [25, 30, 35, 28],
    'City': ['Lahore', 'Karachi', 'Islamabad', 'Lahore'],
    'Score': [88.5, 92.0, 79.5, 95.0]
}
df = pd.DataFrame(data)
df
```

```python
# Concatenating more rows onto the DataFrame
more_data = pd.DataFrame({
    'Name': ['Ethan', 'Farah'], 'Age': [22, 31],
    'City': ['Multan', 'Lahore'], 'Score': [85.0, 90.5]
})
combined = pd.concat([df, more_data], ignore_index=True)
combined
```

## Summary statistics functions

`.describe()` gives a quick table of count, mean, standard deviation, min, max, and quartiles for numeric columns. `.mean()`, `.median()`, `.max()`, `.min()`, and `.std()` compute a single statistic per column. `.value_counts()` counts how often each distinct value appears in a Series, and `.corr()` shows pairwise correlation between numeric columns.

```python
df.describe()
```

```python
df.info()
df.describe(include='all')
```

```python
print("Mean age:", df['Age'].mean())
print("Max score:", df['Score'].max())
print()
print(df['City'].value_counts())
```

## Common methods used very frequently

`.head(n)` / `.tail(n)` preview the first or last rows, `.shape` gives (rows, columns), and `.info()` summarizes column types and non missing counts. `.loc[]` selects by label and `.iloc[]` selects by position. Boolean filtering, for example `df[df['Age'] > 28]`, keeps only rows matching a condition, and `.isna()`, `.dropna()`, `.fillna()` handle missing values.

```python
print(df.head(2))
print()
print("Shape:", df.shape)
```

```python
df.info()
```

```python
print("loc, row 0:\n", df.loc[3])
print()
print("iloc, first two rows and columns:\n", df.iloc[0:1, 0:3])
```

```python
# Boolean filtering
print(df[(df['City'] == 'Lahore') & (df['Score'] > 85)])
```

```python
# Missing data
df_missing = df.copy()
df_missing.loc[1, 'Score'] = np.nan

print("Missing count per column:\n", df_missing.isna().sum())
print()
print("Filled:\n", df_missing.fillna({'Score': df_missing['Score'].mean()}))
```

## Creating new columns and renaming columns

Assigning to a new column name, for example `df['Passed'] = df['Score'] >= 80`, creates a new column, either from a constant, a calculation on existing columns, or a separate list or Series. Pandas computes these values once at that moment, they are not live formulas, so changing the original column later will not update anything derived from it earlier. `.rename(columns={...})` renames columns and returns a new DataFrame unless you pass `inplace=True`.

```python
df2 = df.copy()
df2['Passed'] = df2['Score'] >= 80
df2['Age_in_5_years'] = df2['Age'] + 5
df2
```

```python
# Derived columns are not live formulas
df2['Doubled_age'] = df2['Age'] * 2
df2['Age'] = df2['Age'] + 1
print("Doubled_age did NOT update automatically:")
print(df2[['Age', 'Doubled_age']])
```

```python
renamed = df2.rename(columns={'Name': 'Full_Name', 'Age': 'Current_Age'})
print(renamed.columns.tolist())
```

## Changing the layout and sorting

`.sort_values(by=column)` sorts rows by one or more columns, with `ascending=False` for descending order. `.groupby()`  splits data into groups, applies an aggregation, and combines the results, the classic split, apply, combine pattern.

```python
print(df.sort_values(by='Score', ascending=False))
```

```python
print(df.groupby('City')['Score'].mean())
```

## A few other essential tools

`pd.read_csv()` and `df.to_csv()` read and write CSV files. `pd.merge(left, right, on='key')` joins two DataFrames on a shared column, similar to a SQL join. A single column, `df['Score']`, returns a Series, while a list of columns, `df[['Name', 'Score']]`, returns a DataFrame. `.apply()` runs a custom function across every row or column.

```python
df.to_csv('sample_data.csv', index=False)
pd.read_csv('sample_data.csv')
```

```python
customers = pd.DataFrame({'CustomerID': [1, 2, 3], 'Name': ['Alice', 'Bob', 'Charlie']})
orders = pd.DataFrame({'OrderID': [101, 102, 103], 'CustomerID': [1, 2, 1], 'Amount': [250, 400, 150]})

pd.merge(customers, orders, on='CustomerID')
```

```python
def score_category(score):
    return 'Excellent' if score >= 90 else 'Good' if score >= 80 else 'Needs Improvement'

df3 = df.copy()
df3['Category'] = df3['Score'].apply(score_category)
df3
```
