# Environment and Notebook Basics

## 1. What Each Tool Is

### Python and the interpreter

- **Python**: the programming language, has different versions 
- **Python interpreter**: the program that executes Python code
- One computer can have several interpreters (The interpreter is specific to both a Python version and an environment location)

### Editors and notebooks

- **VS Code**: a code editor or IDE
- **IDE**: integrated development environment, combining editing and development tools
- **Jupyter notebook**: an `.ipynb` file containing Markdown, code, and outputs
- **Jupyter**: a software that gives tools and standards for creating, opening, and interactively running notebook files. also supports the kernels that are actually the runtime/backend that lets you execute code cells in your notebook
- **Google Colab**: a browser-based cloud notebook service by Google 

### Where code runs

- **Local execution**: Python runs on this computer
- **Cloud execution**: Python runs on a remote or cloud-hosted runtime
- Local files and packages are not automatically available in a cloud runtime

## 3. Installation and Environment Verification

### Install

- **Python 3**: installer from [python.org](https://www.python.org/downloads/) (choose the standalone installer)
- **Windows**: Python installer, **Add Python.exe to PATH** 
- **macOS**: Python `.pkg` installer (counterpart of .exe for Windows)

### Terminal check

| Windows | macOS | What they do |
|---|---|---|

### Checking Python from inside the notebook

- `sys` provides information about the running Python
- `sys.version` shows this kernel's Python version 
- when you run the first code cell, you may be prompted to install ipykernel if it is missing, go ahead with it

````python
import sys

print(sys.version)
````

### Checking the active interpreter

- `sys.executable` shows the interpreter's file location
- The result shows the Python executable used by the currently selected notebook kernel, including in Colab


````python
print(sys.executable)
````

## 4. Notebook Anatomy

- **Markdown cells**: explanations and formatted text
- **Code cells**: Python sent to the kernel
- **Outputs**: text, tables, or plots below a code cell
- **Errors**: messages explaining why execution failed

- Double-clicking Markdown reveals its source

## 5. Notebook Execution

### Running cells and keeping values in memory

- Run button or **Shift+Enter**: execute the selected cell
- A **variable** gives a value a name
- The kernel keeps variables and imports in memory across the cells

````python
sample_count = 10
````

### Notebook state

- `sample_count` is available to another cell
- **State**: the variables and imports currently remembered by the kernel

````python
print(sample_count)
````

### Execution order vs visual order

- Code runs in the order cells are executed
- A later assignment changes the value stored in memory

````python
sample_count = 20
````

### Interpreter, environment, and kernel

- **Interpreter**: program that executes Python
- **Environment**: interpreter plus its available packages
- **Kernel**: running process that executes notebook cells and holds state

### Restart and Run All

| Action | Effect |
|---|---|
| Restart | Fresh kernel; variables and imports are cleared |
| Run All | Cells execute top-to-bottom using the current state |
| Restart & Run All | Fresh state, followed by top-to-bottom execution |

- Restart leaves the code and Markdown intact, but old outputs may remain visible


### State after restart

- After restart, `print(sample_count)` raises `NameError` until the defining cell runs
- The defining cell has not run in the new kernel
- Restart & Run All rebuilds state from the notebook cells in order

## 6. Reproducibility Habits

- Top-to-bottom order: imports and assignments before their uses
- Fresh-kernel verification: Restart & Run All
- Final save after verification keeps code and outputs intact

## 7. Python Packages and Imports

### Standard library

- A **module** is a single file that contains reusable code; a **package** is a group of related modules
- `import` makes that code available in the current session
- The standard library comes with Python: no separate installation
- `sys` and `math` are examples of modules

````python
import math

print(math.sqrt(25))
````

### Third-party packages

- NumPy, pandas, and Matplotlib are installed separately from Python
- Colab usually includes these packages already
- Installing makes a package available in an environment; importing uses it in a session

**Terminal: Windows or macOS, with the specific virtual environment active**

```text
python -m pip install numpy pandas matplotlib
```
latest compatible versions available will be installed if you give no version constraints

### Importing and checking NumPy

- `as np`: a short name used inside the notebook
- `np.__version__`: version available to this kernel

````python
import numpy as np

print("NumPy version:", np.__version__)
````

## 8. Virtual Environments

- An environment combines a Python interpreter and its packages
- A virtual environment separates one project's packages from another's
- `.venv` is the usual folder name, you can give it a more meaningful name
,notebooks stay outside it


### Creating and activating: terminal in the working folder

**Windows PowerShell**

```powershell
python -m venv .venv
.\.venv\Scripts\Activate.ps1
```

**macOS**

```bash
python3 -m venv .venv
source .venv/bin/activate
```

- First command: creates `.venv`
- Second command: activates it in the current terminal
- `(.venv)` commonly appears in the prompt
- A new terminal needs activation again if not automatically activated after detection
- Windows alternative for creation: `py -m venv .venv`

**Windows Command Prompt: activation alternative if PowerShell blocks scripts**


"PowerShell cannot run Activate.ps1 because script execution is blocked by the current execution policy."

A fix: 
``` text 
Set-ExecutionPolicy -Scope CurrentUser -ExecutionPolicy RemoteSigned
OR just run the followng in the Windows command prompt terminal
```

```bat
.venv\Scripts\activate.bat
```

### Installing dependencies inside `.venv`

**Terminal: Windows or macOS, after activation**

```text
python -m pip install numpy pandas matplotlib ipykernel
```

- Activated terminals use `.venv`'s `python`
- `ipykernel` lets this environment run notebook cells
- this is an alternative to using the requirements.txt file


### Using `requirements.txt`

A project can list its required packages in a `requirements.txt` file.

**Terminal: activated `.venv`**

```text
python -m pip install -r requirements.txt
```

- `-r` tells `pip` to install the packages listed in the file
- This gives the project one reusable package list instead of installing each package manually

### Kernel Interpreter mismatch

- Notebook: **Select Kernel > Python Environments**, then `.venv`
- Terminal activation alone does not switch an open notebook's kernel
- the venv you created and the kernel you then select should point to the same python executable 


For Conda env, download miniconda from conda's official site (anaconda.com).then follow in a powershell terminal:
```text
conda create -n ai500 python=3.12
conda activate ai500
conda install numpy matplotlib jupyter ipykernel
```
### Local JupyterLab: alternative interface

**Terminal: activated environment**

```text
python -m pip install jupyterlab
python -m jupyterlab
```

- First command: installs JupyterLab in the environment
- Second command: starts the local browser interface; the terminal stays open
- Notebook/Lab kernel selection happens in the browser


### Selecting the environment in JupyterLab

- JupyterLab runs notebooks through a kernel, just like VS Code
- If JupyterLab was installed and started from the active `.venv`, that environment may already appear as a Python kernel
- If `.venv` does not appear, register it as a kernel:

```text
python -m pip install ipykernel
python -m ipykernel install --user --name ai500 --display-name "Python (AI500)"
```

## 11. Basic Troubleshooting

| Error | First question to ask |
|---|---|
| `NameError` | Did the cell defining this name run? |
| `ModuleNotFoundError` | Is the package installed in this kernel's environment? |
| `SyntaxError` | Is the Python syntax valid? |
| `TypeError` | Did the operation receive the expected type? |
| `ValueError` | Is the value valid for this operation? |
| `FileNotFoundError` | Is Python looking in the correct folder? |

### Reading a traceback

- Start with the final error line
- Identify the error type and message
- Move upward to the relevant line of code

```text
Cell In[4], line 1
----> 1 print(missing_score)
NameError: name 'missing_score' is not defined
```

### NameError

- Unknown name: missing definition or spelling mismatch



````python
#print(missing_score)
missing_score = 80
print(missing_score)
````

### ModuleNotFoundError

- `numppy` is a typo; the package name is `numpy`
- Correct spelling but failed import: possible environment mismatch

#### Environment mismatch

A package can be installed successfully but still fail to import if the notebook is using a different Python environment.

- **Python: Select Interpreter** → choose the project's `.venv`
- Notebook **Select Kernel** → choose the same `.venv`
- If `.venv` does not appear as a notebook kernel, activate it and install:

```text
python -m pip install ipykernel
```

**Notebook check**

```python
import sys
print(sys.executable)
```

**Terminal check**

```text
python -c "import sys; print(sys.executable)"
```

Both paths should point inside the same `.venv`. If the kernel was changed, restart it before running the imports again.



````python
# import numppy
import numpy as np

print(np.__version__)
````

### SyntaxError


````python
#print("Vamos"
print("Vamos")
````

### TypeError

- `"3"` is text; `2` is a number
- Numeric addition needs numeric values

````python
# "3" + 2
print('3' + 2)
#np.mean("string"s)
````

### ValueError

- `int("3")` converts numeric text to a whole number
- `int("three")` receives text that cannot be converted

````python
print(int("three"))
#print(int("3"))
````

### FileNotFoundError

- A missing file raises `FileNotFoundError`
- A relative filename is resolved from Python's current directory


**File absent from the current directory**
