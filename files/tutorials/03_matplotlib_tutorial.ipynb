# Matplotlib Tutorial

Matplotlib is the original and most widely used plotting library in Python, and many higher level visualization libraries, including pandas' own plotting, are built on top of it. It gives you fine grained control over every part of a chart, including colors, labels, legends, and layout.

It is used because a good chart reveals trends, outliers, and relationships in data far faster than scanning raw numbers in an array or table. Matplotlib is typically imported through its `pyplot` module, aliased as `plt`.

## Installing and importing Matplotlib

If Matplotlib is not installed, install it with:

```bash
pip install matplotlib
```

We normally import the `pyplot` module and give it the short name `plt`.

We will also use NumPy to generate example data.

````python
import matplotlib.pyplot as plt
import numpy as np
import pandas as pd

print("Matplotlib version:", plt.matplotlib.__version__)
````

## First plot

The simplest Matplotlib plot is a line plot:

```python
plt.plot(x, y)
```

Here:

- `x` contains the horizontal-axis values.
- `y` contains the vertical-axis values.
- Matplotlib plots the corresponding points and connects them with a line.

````python
x = [1, 2, 3, 4, 5]
y = [2, 4, 3, 7, 6]

plt.plot(x, y)
plt.show()
````

### What happens if we only provide `y`?

If only one sequence is provided, Matplotlib automatically uses the positions `0, 1, 2, ...` as the x-values.

````python
values = [10, 20, 15, 30, 25]

plt.plot(values)
plt.show()
````

## Anatomy of a plot

A useful chart normally contains several elements:

- **Title**: describes what the chart shows
- **X-axis label**: explains the horizontal variable
- **Y-axis label**: explains the vertical variable
- **Legend**: identifies multiple data series
- **Grid**: can make values easier to read
- **Ticks**: the values displayed along the axes

````python
x = np.linspace(0, 10, 100)
y = np.sin(x)

plt.plot(x, y, label="sin(x)")

plt.title("Sine Wave")
plt.xlabel("x")
plt.ylabel("sin(x)")
plt.legend()
plt.grid(True)
plt.xticks(
    ticks=[0, np.pi, 2 * np.pi, 3 * np.pi],
    labels=['0', 'π', '2π', '3π']
)

plt.show()
````

## Understanding Figure and Axes

Two important Matplotlib concepts are **Figure** and **Axes**.

- **Figure** = the entire canvas/window containing your visualization.
- **Axes** = one individual plotting area inside the Figure.

For example:

```python
fig, ax = plt.subplots()
```

creates one Figure containing one Axes.

The object-oriented approach is especially useful once you have multiple plots.

````python
fig, ax = plt.subplots()

x = np.linspace(0, 10, 100)
y = np.sin(x)

ax.plot(x, y)
ax.set_title("Sine Wave")
ax.set_xlabel("x")
ax.set_ylabel("sin(x)")

plt.show()
````

### `plt.plot()` vs `ax.plot()`

Both can create the same type of plot.

The `pyplot` approach:

```python
plt.plot(x, y)
plt.title("My Plot")
```

The object-oriented approach:

```python
fig, ax = plt.subplots()
ax.plot(x, y)
ax.set_title("My Plot")
```

For simple one-plot examples, `plt` is convenient. For larger visualizations, the `Axes` approach is usually clearer because you explicitly say **which plot** you are modifying.

## Customizing plots

Matplotlib gives you control over almost every part of a chart.

Common options include:

| Argument | Purpose |
|---|---|
| `color` | Line/marker color |
| `linestyle` | Solid, dashed, dotted, etc. |
| `linewidth` | Line thickness |
| `marker` | Marker shape |
| `markersize` | Marker size |
| `alpha` | Transparency |

You can combine these options in `plot()`.

````python
x = np.linspace(0, 10, 20)
y = np.sin(x)

plt.plot(
    x, y,
    color="steelblue", # can provide hex code for colors also
    linestyle="-", # _ , : , -.
    linewidth=2,
    marker="x",
    markersize=5,
    alpha=0.8
)

plt.title("Customized Line Plot")
plt.xlabel("x")
plt.ylabel("sin(x)")
plt.grid(True)

plt.show()
````

### Common marker styles

Some commonly used markers are:

- `o` → circle
- `s` → square
- `^` → triangle
- `*` → star
- `x` → x marker
- `.` → point

Markers are useful when you want to make individual observations visible.

### Axis limits

You can control the visible range of an axis using `xlim()` and `ylim()`.

````python
x = np.linspace(0, 10, 100)
y = np.sin(x)

plt.plot(x, y)

plt.xlim(0, 10)
plt.ylim(-1.5, 1.5)

plt.title("Controlling Axis Limits")
plt.xlabel("x")
plt.ylabel("sin(x)")

plt.show()
````

### Ticks

Ticks are the values shown along an axis.

You can specify exactly which tick positions should appear.

````python
x = np.linspace(0, 10, 100)
y = np.sin(x)

plt.plot(x, y)

plt.xticks([0, 2, 4, 6, 8, 10])
plt.yticks([-1, -0.5, 0, 0.5, 1])

plt.title("Custom Axis Ticks")
plt.show()
````

For categorical labels or long labels, rotating tick labels can improve readability.

````python
categories = ["January", "February", "March", "April", "May"]
values = [20, 35, 28, 42, 38]

plt.bar(categories, values)
plt.xticks(rotation=35)
plt.ylabel("Sales")
plt.title("Sales by Month")

plt.show()
````

## Multiple data series

You can draw multiple lines on the same Axes. Labels make it possible to identify them using a legend.

````python
x = np.linspace(0, 10, 100)

plt.plot(x, np.sin(x), label="sin(x)")
plt.plot(x, np.cos(x), label="cos(x)")

plt.title("Sine and Cosine")
plt.xlabel("x")
plt.ylabel("value")
plt.legend()
plt.grid(True)

plt.show()
````

## Common plot types

Different charts answer different questions:

| Plot | Useful for |
|---|---|
| Line | Trends and changes |
| Bar | Comparing categories |
| Scatter | Relationships between two variables |
| Histogram | Distribution of one numeric variable |
| Box plot | Comparing distributions and identifying outliers |
| Pie | Showing parts of a whole when there are only a few categories |

The important question is not just **"How do I make this chart?"**, but **"What do I want the chart to communicate?"**

### Line plots

Line plots are especially useful for showing how something changes over an ordered variable, such as time.

````python
months = ["Jan", "Feb", "Mar", "Apr", "May", "Jun"]
sales = [120, 135, 128, 160, 175, 190]

plt.plot(months, sales, marker="o")

plt.title("Monthly Sales")
plt.xlabel("Month")
plt.ylabel("Sales")

plt.show()
````

### Bar charts

Bar charts are useful for comparing separate categories.

````python
categories = ["A", "B", "C", "D"]
values = [23, 45, 12, 36]

plt.bar(categories, values)

plt.title("Category Comparison")
plt.xlabel("Category")
plt.ylabel("Value")

plt.show()
````

For long category names, horizontal bars can be easier to read.

````python
categories = ["Computer Science", "Mathematics", "Physics", "Economics"]
values = [85, 72, 68, 79]

plt.barh(categories, values)

plt.title("Average Scores")
plt.xlabel("Score")

plt.show()
````

### Scatter plots

A scatter plot displays individual observations as points.

It is useful for investigating whether two numeric variables appear to be related.

For example, we can look at the relationship between study hours and exam scores.

````python
rng = np.random.default_rng(seed=0)

study_hours = rng.uniform(1, 10, 100)
exam_scores = 45 + 5 * study_hours + rng.normal(0, 5, 100)

plt.scatter(study_hours, exam_scores, alpha=0.1)

plt.title("Study Hours vs Exam Score")
plt.xlabel("Study Hours")
plt.ylabel("Exam Score")

plt.show()
````

### Histograms

A histogram shows the **distribution** of one numeric variable.

It divides the values into intervals called **bins** and counts how many observations fall into each interval.

The `bins` argument controls how many intervals are used.

````python
rng = np.random.default_rng(seed=1)
data = rng.normal(loc=50, scale=10, size=1000)

plt.hist(data, bins=30, edgecolor="black")

plt.title("Distribution of Scores")
plt.xlabel("Score")
plt.ylabel("Frequency")

plt.show()
````

### Box plots

A box plot summarizes a distribution using statistics such as the median and quartiles. It can also make unusually distant observations visible as potential outliers.

Box plots are particularly useful when comparing several groups.

````python
rng = np.random.default_rng(seed=2)

group_a = rng.normal(50, 5, 200)
group_b = rng.normal(60, 15, 200)

plt.boxplot([group_a, group_b], labels=["Group A", "Group B"])

plt.title("Comparing Two Distributions")
plt.ylabel("Value")

plt.show()
````

### Pie charts

Pie charts show how a total is divided among categories.

They are most readable when there are only a small number of categories. For many categories, a bar chart is usually easier to compare.

````python
sizes = [35, 25, 20, 20]
labels = ["Product A", "Product B", "Product C", "Product D"]

plt.pie(
    sizes,
    labels=labels,
    autopct="%1.1f%%",
    startangle=90
)

plt.title("Market Share")
plt.axis("equal")

plt.show()
````

## Subplots

A Figure can contain multiple Axes.

For example:

```python
fig, axes = plt.subplots(2, 2)
```

creates a 2 × 2 grid containing four plots.

This is useful when several related visualizations need to be displayed together.

````python
fig, axes = plt.subplots(1, 2, figsize=(10, 4))
# 1 row, 2 colums > 2 graphs to display
x = np.linspace(0, 10, 100)

axes[0].plot(x, np.sin(x))
axes[0].set_title("Sine")

axes[1].plot(x, np.cos(x))
axes[1].set_title("Cosine")

plt.tight_layout()
plt.show()
````

### A 2 × 2 grid

When there are two dimensions of subplots, `axes` becomes a 2D array.

For example:

```python
axes[0, 0]  # top-left
axes[0, 1]  # top-right
axes[1, 0]  # bottom-left
axes[1, 1]  # bottom-right
```

````python
fig, axes = plt.subplots(2, 2, figsize=(10, 8))

x = np.linspace(0, 10, 100)
rng = np.random.default_rng(seed=3)

axes[0, 0].plot(x, np.sin(x))
axes[0, 0].set_title("Line Plot")

axes[0, 1].bar(["A", "B", "C"], [10, 24, 15])
axes[0, 1].set_title("Bar Chart")

axes[1, 0].scatter(
    rng.normal(0, 1, 50),
    rng.normal(0, 1, 50),
    alpha=0.7
)
axes[1, 0].set_title("Scatter Plot")

axes[1, 1].hist(rng.normal(50, 10, 500), bins=20)
axes[1, 1].set_title("Histogram")

plt.tight_layout()
plt.show()
````

### Why use `tight_layout()`?

`tight_layout()` automatically adjusts spacing between subplots so that titles, labels, and tick labels are less likely to overlap.

## Adding annotations

Sometimes you want to draw attention to a particular observation.

`annotate()` lets you add text and optionally point an arrow toward a location.

````python
x = np.arange(1, 7)
y = [10, 14, 13, 22, 18, 27]

plt.plot(x, y, marker="o")

plt.annotate(
    "Highest value",
    xy=(6, 27),
    xytext=(4.5, 30),
    arrowprops=dict(arrowstyle="->")
)

plt.title("Annotating a Data Point")
plt.xlabel("Observation")
plt.ylabel("Value")

plt.show()
````

## Plotting directly from pandas

Pandas provides a `.plot()` method for Series and DataFrames.

Under the hood, pandas uses Matplotlib for these plots.

This is convenient when your data is already stored in a DataFrame.

````python
df = pd.DataFrame({
    "Year": [2020, 2021, 2022, 2023, 2024],
    "Revenue": [100, 150, 130, 200, 250]
})

df.plot(
    x="Year",
    y="Revenue",
    kind="line",
    marker="o",
    title="Revenue Over Time"
)

plt.ylabel("Revenue ($k)")
plt.show()
````

You can also use different plot types through the `kind` argument:

```python
df.plot(kind="line")
df.plot(kind="bar")
df.plot(kind="scatter", x="column1", y="column2")
df.plot(kind="hist")
```

Use pandas plotting when you want a quick visualization directly from a DataFrame. Use Matplotlib directly when you need more detailed control over the visualization.

## Saving figures

Use `savefig()` to save a visualization to a file.

It is common to specify:

- `dpi` → image resolution
- `bbox_inches="tight"` → reduces unnecessary whitespace

Call `savefig()` before `show()` when working with the current figure.

````python
x = np.linspace(0, 10, 100)
y = np.sin(x)

plt.plot(x, y)
plt.title("Saved Sine Plot")

plt.savefig(
    "sine_plot.png",
    dpi=150,
    bbox_inches="tight"
)

plt.show()

print("Saved to sine_plot.png")
````
