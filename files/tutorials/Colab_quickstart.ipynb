# Google Colab Quick Start



- The notebook experience is similar to what we saw from VS code environment
- The important difference is **where Python is running**
- In Colab, Python normally runs on a temporary remote machine
- Files on your laptop are **not automatically available** to that machine
- Files can be uploaded temporarily or accessed through Google Drive
- For normal Colab use, we use a hosted runtime. There is almost no environment setup.


## 1. What changes in Colab?

| Local VS Code + Jupyter | Google Colab |
|---|---|
| Python installed on your computer | Python runtime is provided |
| Local `.venv` can isolate packages | Runtime already has its own environment |
| Project folder is on your computer | Runtime has its own remote filesystem |
| Local files persist on your computer | Runtime files may disappear when the session resets |




## 2. Check the Colab runtime

The same checks used locally still tell us **which Python is actually running the notebook**.


```python
import sys

print("Python version:")
print(sys.version)

print("\nPython executable:")
print(sys.executable)
```

### Compare with VS Code

- `sys.version` → Python version in this runtime
- `sys.executable` → interpreter executing this notebook
- The path will look different from a local VS Code setup
- That difference reflects the fact that Colab is running somewhere else


## 3. Where are files stored?

Colab gives the runtime its own filesystem.

A common working location is `/content`.


```python
from pathlib import Path

print("Current working directory:")
print(Path.cwd())
```

### Colab runtime

- In Colab, the remote runtime cannot automatically see files sitting on the laptop
- A file must be **uploaded** or made available through cloud storage such as Google Drive


## 4. Opening an assignment notebook

There are two different ideas here:

- **Open an `.ipynb` notebook** → the notebook itself becomes the document being edited in Colab
- **Upload a data/supporting file like reqs txt** → the file becomes available to code running inside the notebook

For an assignment notebook already saved on the computer, Colab's **File → Upload notebook** option can open the `.ipynb` file directly.



### Important limitation

- The direct file upload is convenient for a short session
- It is **not reliable long-term storage**
- Resetting or replacing the runtime can remove session files

For the kind of assignment work that should persist across sessions, Google Drive is usually more convenient.


## 5. Mount Google Drive

Mounting Drive makes files in Google Drive accessible from the Colab runtime.

This is useful when:

- assignment notebooks or datasets should persist across sessions
- generated outputs need to be saved somewhere persistent


```python
from google.colab import drive
drive.mount('/content/drive')
```

### Drive location

After mounting, **My Drive** is normally available under:

`/content/drive/MyDrive`


## 7. A dedicated data folder in Drive

You can create a dedicated folder in your Google Drive that gives notebooks and supporting files a stable location. You may upload your datasets for example in this folder.

Example path:

`MyDrive/AI500`


```python
course_folder = Path("/content/drive/MyDrive/AI500")


print(course_folder.exists())

```

If that folder already exists, files can be read from it directly.

A folder can also be created once when needed:


```python
# if the folder does not already exist
course_folder.mkdir(parents=True, exist_ok=True)
```

## 8. Practical assignment workflow in Colab

A compact workflow for working on course assignments may be:

- Open the assignment notebook `.ipynb` in Colab
- Keep persistent datasets/supporting files in a dedicated Drive folder
- Mount Drive when those files are needed
- Use runtime upload for quick temporary files
- Check `sys.executable` if you want to confirm that package installation and notebook execution are happening in the same Python environment
- Save important outputs back to Drive rather than relying on runtime storage

### The following stays the same from local Jupyter

- Code cells
- Markdown cells
- Kernel/runtime state
- Execution order
- Imports
- Errors and tracebacks
- Restart / Run All

Only the **location of the environment and files** changes.
